```python
import pandas as pd

def clean_water_level(row):
    val_str = str(row["Water Level (mMSL)"])
    sl_str = str(int(row["Sl."]))
    if val_str.endswith(sl_str) and len(val_str) > len(sl_str):
        # Remove the serial number from the end of the decimal
        val_str = val_str[: -len(sl_str)]
    return float(val_str)


url = "http://www.hydrology.bwdb.gov.bd/includes/get_water_level_datatable.php?station_id=SW49"

try:
    # Read all tables from the page
    tables = pd.read_html(url)
    print(tables)
    # Usually, the data is in the last table on this specific site
    df = tables[0]

    df["Water Level (mMSL)"] = df.apply(clean_water_level, axis=1)

    # 2. Reformat the Date to YYYY-MM-DD
    # This converts "1 December, 2025" to "2025-12-01"
    df["Date"] = pd.to_datetime(df["Date"]).dt.strftime('%Y-%m-%d')

    
    # Save to CSV
    df.to_csv("bwdb_water_level.csv", index=False)
    print("CSV file 'bwdb_water_level.csv' has been created successfully.")
except Exception as e:
    print(f"Error: {e}. (Note: You may need to provide specific form parameters for some stations).")
```
